## Hour 13 - Lists fundamentals


````python
roster = ["Ava", "Ben", "Cara", "Dev", "Eli"]
original_size = len(roster)
first_learner = roster[0]
last_learner = roster[-1]
middle_learners = roster[1:-1]

roster[2] = "Cora"
roster.append("Finn")

print(f"Roster size: {original_size}")
print(f"First learner: {first_learner}")
print(f"Last learner: {last_learner}")
print(f"Middle learners: {middle_learners}")
print(f"Updated roster: {roster}")
````

## Hour 14 - Iterating lists with for-loops


````python
names = ["Anna", "Christopher", "Maya", "Leo", "Chloe"]
total_characters = 0
longest_name = ""
for name in names:
    total_characters += len(name)
    if len(name) > len(longest_name):
        longest_name = name

average_length = total_characters / len(names)
starts_with_c = [name for name in names if name.startswith("C")]
uppercase_roster = [name.upper() for name in ["Anna", "Chris", "Maya", "Leo"]]

print(f"Name count total: {total_characters}")
print(f"Longest name: {longest_name}")
print(f"Average length: {average_length:.2f}")
print(f"Starts with C: {starts_with_c}")
print(f"Uppercase roster: {uppercase_roster}")
````

## Hour 15 - Nested lists (table-like data)


````python
seats = [
    ["A1", "A2", "A3"],
    ["B1", "B2", "B3"],
    ["C1", "C2", "C3"],
]
flattened = [seat for row in seats for seat in row]

print(f"Rows: {len(seats)}")
print(f"Columns per row: {len(seats[0])}")
print(f"Seat at row 2 col 3: {seats[1][2]}")
print(f"Flattened seats: {','.join(flattened)}")
print(f"Total seats: {len(flattened)}")
````
