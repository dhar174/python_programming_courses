## Part 1: Hour 37 - Intro classes, objects, attributes, methods


### Exercise 1.1: Build a `Contact` class (graded)

Create a Basics-friendly class that matches the runbook lab:

1. Define `Contact` with `__init__(name, phone)`.
2. Store `name` and `phone` as instance attributes.
3. Add `display()` that returns `"Name -> phone"`.
4. Add `announce_ready()` that returns `"Name is ready to contact."`.
5. Instantiate `Ada` with phone `5551234` and use method calls in your graded demo.


````python
starter_name = "Ada"
starter_phone = "5551234"

class Contact:
    """Store one contact with a name and phone number."""

    def __init__(self, name: str, phone: str) -> None:
        self.name = name
        self.phone = phone

    def display(self) -> str:
        return f"{self.name} -> {self.phone}"

    def ready_message(self) -> str:
        return f"{self.name} is ready to contact."

starter_contact = Contact(starter_name, starter_phone)

print("Class defined: Contact(name, phone)")
print(f"Object created: {starter_contact.name} / {starter_contact.phone}")
print(f"Display method: {starter_contact.display()}")
print(f"Method call result: {starter_contact.ready_message()}")
print("self reminder: self refers to the current object")
````

### Exercise 1.2: Create more than one object (practice)

Create at least one more `Contact` object and compare the outputs of `display()`.
Notice that each object stores its own attribute values.


````python
second_contact = Contact("Grace", "5552222")
third_contact = Contact("Linus", "5559999")
practice_cards = [starter_contact.display(), second_contact.display(), third_contact.display()]
assert practice_cards[1] == "Grace -> 5552222"
````

## Part 2: Hour 38 - Collections of objects and searching


### Exercise 2.1: Search a list of contacts (graded)

Refactor the contact manager idea from dictionary records to objects:

1. Build a list containing `Ada`, `Grace`, and `Linus` `Contact` objects.
2. Write `normalize_name(name)` so searching ignores case and outer spaces.
3. Write `find_contact_by_name(contacts, target_name)` that returns the matching object or `None`.
4. Search for `"grace"` and prove it matches `Grace`.
5. Update the found contact's phone to `5553333`.
6. Print the object count to show the collection still contains three records.


````python
target_lookup = "grace"
missing_lookup = "Maya"

def normalize_name(name: str) -> str:
    return name.strip().lower()

def find_contact(contacts: list[Contact], target_name: str) -> Contact | None:
    normalized_target = normalize_name(target_name)
    for contact in contacts:
        if normalize_name(contact.name) == normalized_target:
            return contact
    return None

contacts = [starter_contact, second_contact, third_contact]
found_contact = find_contact(contacts, target_lookup)
missing_contact = find_contact(contacts, missing_lookup)
found_before_update = found_contact.display() if found_contact is not None else "None"
if found_contact is not None:
    found_contact.phone = "5553333"

print(f"Found contact: {found_before_update}")
print(f"Missing contact: {missing_contact}")
print("Search uses normalize: 'grace' matched 'Grace'")
print(f"Updated phone: {found_contact.display()}")
print(f"Object count: {len(contacts)}")
````

### Exercise 2.2: Show contact cards with a loop (practice)

Write a helper that loops through a list of `Contact` objects and returns a list of display strings.
This is good practice for menu-driven programs that need readable list output.


````python
def list_contact_cards(contacts: list[Contact]) -> list[str]:
    return [contact.display() for contact in contacts]

contact_cards = list_contact_cards(contacts)
assert contact_cards[0] == "Ada -> 5551234"
assert contact_cards[1] == "Grace -> 5553333"
````

## Part 3: Hour 39 - Basic encapsulation and data validation


### Exercise 3.1: Keep validation inside the class (graded)

Extend your `Contact` class so business rules stay in one place:

1. Add `update_phone(new_phone)` to `Contact`.
2. Accept a new phone only if it contains digits only and has length 7.
3. Return `True` when the update succeeds.
4. Return `False` when the update is invalid and keep the previous phone unchanged.
5. Test with `"5554444"` and then with `"55-abc"`.
6. Write one short explanation for why validation belongs in the class instead of only in UI code.


````python
valid_phone = "5554444"
invalid_phone = "55-abc"

class Contact:
    def __init__(self, name: str, phone: str) -> None:
        self.name = name
        self.phone = phone

    def display(self) -> str:
        return f"{self.name} -> {self.phone}"

    def update_phone(self, new_phone: str) -> bool:
        candidate = new_phone.strip()
        if candidate.isdigit() and len(candidate) == 7:
            self.phone = candidate
            return True
        return False

def validation_explanation() -> str:
    return "class methods keep business rules together"

validated_contact = Contact("Grace", "5553333")
valid_update_result = validated_contact.update_phone(valid_phone)
stored_after_valid = validated_contact.phone
invalid_update_result = validated_contact.update_phone(invalid_phone)

print(f"Valid update accepted: {valid_update_result}")
print(f"Stored phone after valid update: {stored_after_valid}")
print(f"Invalid update accepted: {invalid_update_result}")
print(f"Stored phone after invalid update: {validated_contact.phone}")
print(f"Validation location: {validation_explanation()}")
````

### Exercise 3.2: Optional extension - normalize digits for storage (practice)

Optional: strip spaces before validation so input like `" 5554444 "` can still become a clean stored phone number.
Keep the rule simple and stay inside Basics scope.


````python
raw_phone_inputs = [" 5554444 ", "555-4444", "abc"]

def normalize_phone(raw_phone: str) -> str:
    return raw_phone.strip()

phone_results = {
    raw: normalize_phone(raw) if normalize_phone(raw).isdigit() else None
    for raw in raw_phone_inputs
}
assert phone_results[" 5554444 "] == "5554444"
assert phone_results["555-4444"] is None
````
