## Part 1: Hour 29 - Menu loops (CLI pattern)


### Exercise 1.1: Menu Loop Routing (graded)

Create a loop-driven CLI simulation for options `1-4` with this deterministic input sequence:
`["1", "2", "3", "4"]`

Print these exact lines (in order):
- `Menu display: [1] Add [2] List [3] Search [4] Quit`
- `Menu option selected: 1 -> add`
- `Menu option selected: 2 -> list`
- `Menu option selected: 3 -> search`
- `Menu option selected: 4 -> quit`
- `Menu loop ended: user requested quit`


````python
menu_inputs = ["1", "2", "3", "4"]
actions = {"1": "add", "2": "list", "3": "search", "4": "quit"}

print("Menu display: [1] Add [2] List [3] Search [4] Quit")
index = 0
while index < len(menu_inputs):
    choice = menu_inputs[index]
    action = actions.get(choice, "invalid")
    print(f"Menu option selected: {choice} -> {action}")
    index += 1
    if choice == "4":
        break
print("Menu loop ended: user requested quit")
````

### Exercise 1.2: Invalid Choice Guard (practice)

Extend your menu loop so invalid choices (for example `"9"`) print a helpful message and continue prompting.
Use deterministic pseudo-input data in code (no real-time interactive input required).


````python
menu_inputs_with_invalid = ["9", "2", "4"]
invalid_choice_demo: list[str] = []
for choice in menu_inputs_with_invalid:
    if choice not in {"1", "2", "3", "4"}:
        invalid_choice_demo.append(f"Invalid choice handled: {choice}")
        continue
    invalid_choice_demo.append(f"Valid choice handled: {choice}")
    if choice == "4":
        break

assert invalid_choice_demo[0] == "Invalid choice handled: 9"
````

## Part 2: Hour 30 - Input validation (Basics approach)


### Exercise 2.1: Safe Integer Entry (graded)

Use this deterministic sequence to simulate user input:
`["abc", "-5", "12"]`

Apply a basic `.isdigit()` guard and print these exact lines:
- `Input 'abc' -> invalid integer`
- `Input '-5' -> invalid for isdigit guard`
- `Input '12' -> accepted integer 12`
- `Validated number doubled: 24`
- `Validation note: isdigit() does not accept negatives or decimals`


````python
raw_inputs = ["abc", "-5", "12"]
validated_number: int | None = None
validation_messages: list[str] = []
index = 0

while validated_number is None and index < len(raw_inputs):
    raw = raw_inputs[index]
    index += 1
    if raw.isdigit():
        validated_number = int(raw)
        validation_messages.append(f"Input '{raw}' -> accepted integer {validated_number}")
    elif raw.startswith("-") and raw[1:].isdigit():
        validation_messages.append(f"Input '{raw}' -> invalid for isdigit guard")
    else:
        validation_messages.append(f"Input '{raw}' -> invalid integer")

for message in validation_messages:
    print(message)
print(f"Validated number doubled: {validated_number * 2}")
print("Validation note: isdigit() does not accept negatives or decimals")
````

### Exercise 2.2: Optional Extension - Negative Integers (practice)

Optional: support negatives by checking `raw.lstrip('-').isdigit()` after confirming `raw` is not just `"-"`.
Document one limitation that still remains for decimal values.


````python
optional_inputs = ["-12", "3.5", "7"]

def parse_integer_with_negatives(raw: str) -> int | None:
    stripped = raw.strip()
    if stripped.startswith("-"):
        return -int(stripped[1:]) if stripped[1:].isdigit() else None
    return int(stripped) if stripped.isdigit() else None

optional_results = {raw: parse_integer_with_negatives(raw) for raw in optional_inputs}
assert optional_results["-12"] == -12
assert optional_results["3.5"] is None
````

## Part 3: Hour 31 - Mini-project: CLI Contact Manager (in-memory)


### Exercise 3.1: Contact Manager Core Features (graded)

Build a menu-style in-memory Contact Manager with required actions:
- Add contact (name, phone)
- List contacts
- Search by name
- Delete by name

For deterministic grading, print these exact lines:
- `Added contact: Ava -> 555-0101`
- `Added contact: Ben -> 555-0102`
- `List contacts: Ava (555-0101), Ben (555-0102)`
- `Search 'Ava': 555-0101`
- `Delete 'Ben': removed`
- `Delete 'Zoe': not found`


````python
contacts: dict[str, str] = {}

def add_contact(name: str, phone: str) -> str:
    contacts[name] = phone
    return f"Added contact: {name} -> {phone}"

def list_contacts() -> str:
    return ", ".join(f"{name} ({phone})" for name, phone in contacts.items())

def search_contact(name: str) -> str:
    return contacts.get(name, "not found")

def delete_contact(name: str) -> str:
    if name in contacts:
        del contacts[name]
        return "removed"
    return "not found"

print(add_contact("Ava", "555-0101"))
print(add_contact("Ben", "555-0102"))
print(f"List contacts: {list_contacts()}")
print(f"Search 'Ava': {search_contact('Ava')}")
print(f"Delete 'Ben': {delete_contact('Ben')}")
print(f"Delete 'Zoe': {delete_contact('Zoe')}")
````

### Exercise 3.2: UX and Data Quality (practice)

Practice clean UX messaging for duplicate names and case sensitivity.
Decide and document one policy for handling `"ava"` vs `"Ava"`.


````python
sample_names = ["Ava", "ava", "AVA"]

def normalize_name(name: str) -> str:
    return name.strip().title()

normalized_names = [normalize_name(name) for name in sample_names]
unique_normalized_names = sorted(set(normalized_names))
assert unique_normalized_names == ["Ava"]
````

## Part 4: Hour 32 - Checkpoint 4: Control flow assessment


### Exercise 4.1: CLI To-Do Manager Checkpoint (graded)

Build an in-memory To-Do Manager with:
- Menu loop until quit
- Add task
- List tasks
- Mark complete
- Delete task
- Basic input validation (non-empty task)

For deterministic grading, print these exact lines:
- `Task added: Finish checkpoint`
- `Task added: Review loops`
- `Task completed: Finish checkpoint`
- `Task deleted: Review loops`
- `Remaining tasks: Finish checkpoint [done]`
- `Checkpoint 4 status: pass`


````python
tasks: list[dict[str, object]] = []

def add_task(title: str) -> str:
    tasks.append({"title": title, "done": False})
    return f"Task added: {title}"

def complete_task(title: str) -> str:
    for task in tasks:
        if task["title"] == title:
            task["done"] = True
            return f"Task completed: {title}"
    return f"Task missing: {title}"

def delete_task(title: str) -> str:
    for index, task in enumerate(tasks):
        if task["title"] == title:
            tasks.pop(index)
            return f"Task deleted: {title}"
    return f"Task missing: {title}"

def list_tasks() -> str:
    return ", ".join(
        f"{task['title']} [{'done' if task['done'] else 'open'}]"
        for task in tasks
    )

print(add_task("Finish checkpoint"))
print(add_task("Review loops"))
print(complete_task("Finish checkpoint"))
print(delete_task("Review loops"))
print(f"Remaining tasks: {list_tasks()}")
print("Checkpoint 4 status: pass")
````
