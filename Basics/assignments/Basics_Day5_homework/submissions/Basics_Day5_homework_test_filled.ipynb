## Hour 17 - Tuples + unpacking


````python
coordinates = (12, 34)
x_value, y_value = coordinates
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri")
roles = ("mentor", "student")
updated_roles = roles + ("reviewer",)

print(f"Coordinates tuple: {coordinates}")
print(f"X value: {x_value}")
print(f"Y value: {y_value}")
print(f"First three days: {weekdays[:3]}")
print(f"Updated roles: {updated_roles}")
````

## Hour 18 - Sets: uniqueness + membership


````python
colors = ["blue", "green", "blue", "red", "violet", "green", "blue"]
unique_colors = set(colors)
sorted_unique_colors = sorted(unique_colors)

print(f"Raw item count: {len(colors)}")
print(f"Unique item count: {len(unique_colors)}")
print(f"Is 'blue' present: {'blue' in unique_colors}")
print(f"Is 'yellow' present: {'yellow' in unique_colors}")
print(f"Unique sorted colors: {sorted_unique_colors}")
````

## Hour 19 - Dictionaries fundamentals


````python
learner_profile = {"name": "Nia", "role": "analyst", "city": "Accra"}
learner_profile["city"] = "Lagos"
sorted_keys = sorted(learner_profile)

print(f"Profile keys: {sorted_keys}")
print(f"Learner name: {learner_profile['name']}")
print(f"Learner role: {learner_profile['role']}")
print(f"Updated city: {learner_profile['city']}")
print(f"Has 'email' key: {'email' in learner_profile}")
````

## Hour 20 - Dictionary iteration + counting pattern


````python
words = ["python", "rocks", "data", "python", "fun", "data", "python", "rocks"]
word_counts: dict[str, int] = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

summary = ", ".join(f"{word}={word_counts[word]}" for word in sorted(word_counts))
print(f"Total words: {len(words)}")
print(f"Unique words: {len(word_counts)}")
print(f"Count for 'python': {word_counts['python']}")
print(f"Count for 'data': {word_counts['data']}")
print(f"Word counts: {summary}")
````
