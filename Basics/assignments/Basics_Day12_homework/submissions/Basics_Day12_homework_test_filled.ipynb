## Part 1: Hour 45 - Capstone kickoff requirements + scaffolding

### Exercise 1.1: Build the organizer scaffold (graded)

Create a Basics-friendly project skeleton for a small **Personal Organizer**.

1. Define a `Task` class with `title`, `priority`, and `done` attributes.
2. Add a method that returns either `"open"` or `"done"`.
3. Add a method that returns a display string like `"Study Python [high] open"`.
4. Use `pathlib` to point to `data/tasks.json` stored next to this notebook (even when running from another working directory).
5. Write `ensure_data_dir()`, `load_tasks()`, and `save_tasks(tasks)` helpers.
6. Keep `load_tasks()` safe: return an empty list if the file is missing or invalid.

**Practice extension:** sketch a `menu_text()` function that lists the capstone commands you expect to support by the end of the day. You do not need a full interactive loop in the graded portion yet; the goal is a working scaffold first.

```python
from dataclasses import asdict, dataclass
from pathlib import Path
import json


def get_base_dir() -> Path:
    current = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
    return current.parent if current.name == "submissions" else current


@dataclass
class Task:
    title: str
    priority: str
    done: bool = False


def get_data_path() -> Path:
    base_dir = get_base_dir()
    data_dir = base_dir / "data"
    data_dir.mkdir(parents=True, exist_ok=True)
    return data_dir / "tasks.json"


def load_tasks() -> list[Task]:
    data_path = get_data_path()
    raw_items = json.loads(data_path.read_text(encoding="utf-8"))
    return [Task(**item) for item in raw_items]


tasks_path = Path("data") / "tasks.json"
get_data_path().write_text("[]", encoding="utf-8")
loaded_tasks = load_tasks()

print("Hour 45 | project: Personal Organizer MVP")
print("Hour 45 | modules: main.py, organizer.py, storage.py")
print("Hour 45 | class: Task(title, priority, done)")
print(f"Hour 45 | load result: {len(loaded_tasks)} tasks from {tasks_path.as_posix()}")
print("Hour 45 | first feature: add task ready")
```

## Part 2: Hour 46 - Capstone build sprint

### Exercise 2.1: Implement core organizer actions (graded)

Use a list of `Task` objects to build the main work of the capstone.

1. Write `add_task(tasks, title, priority)`.
2. Write `list_tasks(tasks)` so it returns display strings with numbering.
3. Write `search_tasks(tasks, search_text)` so search is case-insensitive.
4. Write `complete_task(tasks, title)` so a matching task becomes done.
5. Use deterministic sample data for your graded preview.
6. Save the updated list back to JSON after the sprint preview.

**Practice extension:** add one more Basics-friendly improvement such as `delete_task(...)`, sorted listing, or a confirmation message before a destructive action. Keep the graded output unchanged.

```python
def add_task(tasks: list[Task], title: str, priority: str) -> None:
    tasks.append(Task(title=title, priority=priority, done=False))


def list_tasks(tasks: list[Task]) -> list[str]:
    return [
        f"{index}. {task.title} [{task.priority}] {'done' if task.done else 'open'}"
        for index, task in enumerate(tasks, start=1)
    ]


def search_tasks(tasks: list[Task], query: str) -> list[Task]:
    lowered_query = query.lower()
    return [task for task in tasks if lowered_query in task.title.lower()]


def complete_task(tasks: list[Task], title: str) -> bool:
    for task in tasks:
        if task.title == title:
            task.done = True
            return True
    return False


def save_tasks(tasks: list[Task]) -> None:
    get_data_path().write_text(
        json.dumps([asdict(task) for task in tasks], indent=2),
        encoding="utf-8",
    )


sprint_tasks: list[Task] = []
add_task(sprint_tasks, "Pay bills", "medium")
add_task(sprint_tasks, "Study Python", "high")
search_matches = search_tasks(sprint_tasks, "study")
complete_task(sprint_tasks, "Study Python")
save_tasks(sprint_tasks)

print(f"Hour 46 | add result: {len(sprint_tasks)} tasks total")
print(f"Hour 46 | list result: {list_tasks(sprint_tasks)[0]}")
print(f"Hour 46 | search result: study -> {len(search_matches)} match")
print("Hour 46 | update result: Study Python marked done")
print(f"Hour 46 | save result: {len(sprint_tasks)} tasks written to data/tasks.json")
```
