## Part 1: Hour 33 - REST fundamentals + Flask app setup

**Goal:** Build a Flask app with a health route and clear REST basics.

**Best practice:** Start with one small health endpoint before layering CRUD.

**Avoid this pitfall:** Mixing setup, routing, and service code in one giant file is hard to scale.


````python
class FakeFlaskApp:
    def __init__(self, name: str) -> None:
        self.name = name
        self.routes: dict[tuple[str, str], callable] = {}

    def route(self, path: str, methods: list[str]):
        def decorator(func):
            for method in methods:
                self.routes[(method, path)] = func
            return func
        return decorator

app = FakeFlaskApp("tracker")

@app.route("/health", methods=["GET"])
def health() -> tuple[dict[str, str], int]:
    return {"status": "ok"}, 200

payload, status_code = app.routes[("GET", "/health")]()

print("Hour 33 | framework: Flask")
print("Hour 33 | route: GET /health")
print(f"Hour 33 | status code: {status_code}")
print(f"Hour 33 | json response: {payload}")
print("Hour 33 | rest rule: name resources clearly")
````

## Part 2: Hour 34 - CRUD endpoints for your main resource

**Goal:** Build a small set of CRUD endpoints for the tracker resource.

**Best practice:** Use HTTP verbs and status codes consistently across CRUD routes.

**Avoid this pitfall:** Returning the wrong status code makes clients harder to trust.


````python
tasks_api_store: dict[str, dict[str, object]] = {}

def create_task(task_id: str, title: str, priority: str) -> int:
    tasks_api_store[task_id] = {"id": task_id, "title": title, "priority": priority, "done": False}
    return 201

def list_tasks() -> int:
    return 200

def delete_task(task_id: str) -> int:
    tasks_api_store.pop(task_id, None)
    return 204

create_status = create_task("task-101", "Ship report", "high")
list_status = list_tasks()
delete_status = delete_task("task-101")

print("Hour 34 | endpoint: POST /tasks")
print(f"Hour 34 | create status: {create_status}")
print(f"Hour 34 | list status: {list_status}")
print(f"Hour 34 | delete status: {delete_status}")
print("Hour 34 | endpoint rule: align verbs with actions")
````

## Part 3: Hour 35 - Serialization + validation (manual, consistent)

**Goal:** Build manual serializers and validators for tracker API payloads.

**Best practice:** Validate incoming payloads and serialize responses with one consistent contract.

**Avoid this pitfall:** Returning different key sets across endpoints confuses clients.


````python
def validate_task_payload(payload: dict[str, object]) -> tuple[bool, str]:
    if not str(payload.get("title", "")).strip():
        return False, "title is required"
    if "priority" not in payload:
        return False, "priority is required"
    return True, "ok"

def serialize_task(task: dict[str, object]) -> dict[str, object]:
    return {key: task[key] for key in ["id", "title", "priority", "done"]}

invalid_payload = {"title": "   "}
valid_task = {"id": "task-101", "title": "Ship report", "priority": "high", "done": False}
is_valid, validation_message = validate_task_payload(invalid_payload)
serialized_task = serialize_task(valid_task)

print("Hour 35 | request fields checked: title, priority")
print(f"Hour 35 | invalid payload: {400 if not is_valid else 200}")
print(f"Hour 35 | serialized task keys: {', '.join(serialized_task)}")
print(f"Hour 35 | validation message: {validation_message}")
print("Hour 35 | contract rule: keep input and output shapes consistent")
````

## Part 4: Hour 36 - App structure: blueprints (optional) + dependency wiring

**Goal:** Build an app structure that can grow beyond one file.

**Best practice:** Use an app factory and optional blueprints to keep startup wiring organized.

**Avoid this pitfall:** Global singletons that hide dependencies make testing harder.


````python
class DevConfig:
    DEBUG = True

class Blueprint:
    def __init__(self, prefix: str) -> None:
        self.prefix = prefix

class TaskService:
    def __init__(self) -> None:
        self.name = "task_service"

def create_app() -> dict[str, object]:
    return {
        "config": DevConfig,
        "blueprint": Blueprint("/api"),
        "task_service": TaskService(),
    }

built_app = create_app()

print("Hour 36 | app factory: create_app")
print(f"Hour 36 | blueprint prefix: {built_app['blueprint'].prefix}")
print(f"Hour 36 | dependency wired: {built_app['task_service'].name}")
print(f"Hour 36 | config class: {built_app['config'].__name__}")
print("Hour 36 | structure rule: keep startup wiring separate from route logic")
````
