## Part 3: Hour 39 - Integration choice: GUI talks to API OR parallel UI/API

**Goal:** Build executable integration behavior that keeps the API as the source of truth.

**Definition of Done (used by grading):**
1. Implement `submit_task_via_api(api_client, title)`.
2. Success path returns a dict with `status=201` and `source='api'`.
3. Failure path catches connection failure and returns `status=503` with an `api unavailable` message.
4. Write `integration_artifacts/hour39_integration_trace.json` showing both success and failure attempts.

**Best practice:** Pick one integration path and document how components communicate.

**Avoid this pitfall:** Running parallel flows without a clear source of truth creates drift.
            


```python
# Hour 39 starter
# Focus: Integration choice: GUI talks to API OR parallel UI/API
# Deliverables checked by grading: submit_task_via_api behavior + integration trace artifact

import json
from pathlib import Path


def submit_task_via_api(api_client, title: str) -> dict:
    try:
        response = api_client.create_task(title)
        status = int(response.get("status", 201)) if isinstance(response, dict) else 201
        return {"status": status, "source": "api", "message": "created", "fallback": "none"}
    except ConnectionError as exc:
        return {"status": 503, "source": "api", "message": f"api unavailable: {exc}", "fallback": "none"}


class DemoSuccessClient:
    def create_task(self, title: str) -> dict:
        return {"status": 201, "task": {"title": title}}


class DemoFailingClient:
    def create_task(self, title: str) -> dict:
        raise ConnectionError("api unavailable")


integration_trace = [
    {"step": "success_create", **submit_task_via_api(DemoSuccessClient(), "Review API client")},
    {"step": "failure_create", **submit_task_via_api(DemoFailingClient(), "Review API client")},
]

artifact_dir = Path("integration_artifacts")
artifact_dir.mkdir(parents=True, exist_ok=True)
(artifact_dir / "hour39_integration_trace.json").write_text(json.dumps(integration_trace, indent=2), encoding="utf-8")
            
```
