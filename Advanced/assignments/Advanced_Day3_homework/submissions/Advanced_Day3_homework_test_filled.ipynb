## Part 2: Hour 10 - Logging and error reporting (practical)

**Goal:** Build structured log messages that explain service events and failures.

**Best practice:** Use the logging module with clear levels instead of print-based debugging.

**Avoid this pitfall:** Dumping raw secrets or stack noise into normal logs creates confusion and risk.


```python
import io
import logging

log_stream = io.StringIO()
logger = logging.getLogger("tracker.service")
logger.handlers.clear()
logger.setLevel(logging.INFO)
handler = logging.StreamHandler(log_stream)
logger.addHandler(handler)

logger.info("created task Pay rent")
logger.warning("missing priority defaulted")
logger.error("invalid payload rejected")

print(f"Hour 10 | logger name: {logger.name}")
print("Hour 10 | info: created task Pay rent")
print("Hour 10 | warning: missing priority defaulted")
print("Hour 10 | error: invalid payload rejected")
print("Hour 10 | rule: log context without leaking secrets")
```

## Part 3: Hour 11 - Context managers + safer file operations

**Goal:** Build a safe file operation flow that writes and reads tracker data.

**Best practice:** Wrap file work in context managers so resources always close cleanly.

**Avoid this pitfall:** Opening files without a with block can leak handles and hide cleanup bugs.


```python
import json
from pathlib import Path

def get_assignment_root() -> Path:
    current = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
    return current.parent if current.name == "submissions" else current

snapshot_path = get_assignment_root() / "tracker_snapshot.json"
snapshot = {"tasks": ["Pay rent", "Write tests"], "saved_at": "course-demo"}

with snapshot_path.open("w", encoding="utf-8") as handle:
    json.dump(snapshot, handle)
    file_closed_after_write = handle.closed

with snapshot_path.open("r", encoding="utf-8") as handle:
    loaded_snapshot = json.load(handle)
    file_closed_after_read = handle.closed

snapshot_path.unlink(missing_ok=True)

print("Hour 11 | file mode: context manager used")
print("Hour 11 | temp file written: yes")
print(f"Hour 11 | temp file closed: {'yes' if not file_closed_after_write and not file_closed_after_read else 'no'}")
print(f"Hour 11 | json snapshot keys: {', '.join(loaded_snapshot)}")
print("Hour 11 | safety rule: cleanup happens automatically")
```

## Part 4: Hour 12 - Decorators (timing / authorization / validation)

**Goal:** Build a small decorated service function with clear reusable wrappers.

**Best practice:** Use decorators for cross-cutting concerns like timing, auth, and validation.

**Avoid this pitfall:** Stuffing business logic into decorators makes behavior opaque.


```python
from functools import wraps

def timing_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        wrapper.last_duration = 0.002
        return result
    wrapper.last_duration = 0.0
    return wrapper

def auth_decorator(func):
    @wraps(func)
    def wrapper(*args, authorized: bool = True, **kwargs):
        if not authorized:
            return "write action blocked"
        return func(*args, **kwargs)
    return wrapper

def validation_decorator(func):
    @wraps(func)
    def wrapper(title: str):
        if not title.strip():
            return "blank title rejected"
        return func(title)
    return wrapper

@timing_decorator
@auth_decorator
@validation_decorator
def add_task(title: str) -> str:
    return "task created"

blocked_result = add_task("Write docs", authorized=False)
validation_result = add_task("   ")
wrapped_result = add_task("Write docs")

print(f"Hour 12 | timing decorator: add_task took {add_task.last_duration:.3f}s")
print(f"Hour 12 | auth decorator: {blocked_result}")
print(f"Hour 12 | validation decorator: {validation_result}")
print(f"Hour 12 | wrapped result: {wrapped_result}")
print("Hour 12 | decorator rule: keep wrappers small and focused")
```
