```python
from dataclasses import asdict, dataclass
import json
from pathlib import Path


@dataclass
class TaskItem:
    task_id: str
    title: str


class TaskService:
    def __init__(self) -> None:
        self._tasks: dict[str, TaskItem] = {}

    def add_task(self, task_id: str, title: str) -> None:
        normalized = title.strip()
        if not normalized:
            raise ValueError("title must not be blank")
        self._tasks[task_id] = TaskItem(task_id=task_id, title=normalized)

    def get_task(self, task_id: str) -> TaskItem:
        return self._tasks[task_id]

    def list_tasks(self) -> list[TaskItem]:
        return list(self._tasks.values())


class TaskWorkflowService(TaskService):
    def add_unique_task(self, task_id: str, title: str) -> None:
        if task_id in self._tasks:
            raise ValueError("duplicate id rejected")
        self.add_task(task_id, title)

    def search(self, term: str) -> list[TaskItem]:
        lowered = term.lower()
        return [task for task in self._tasks.values() if lowered in task.title.lower()]


def write_artifacts(service: TaskWorkflowService) -> None:
    data_dir = Path("data")
    data_dir.mkdir(parents=True, exist_ok=True)
    payload = [asdict(task) for task in service.list_tasks()]
    (data_dir / "task_snapshot.json").write_text(json.dumps(payload, indent=2), encoding="utf-8")

    packaging_root = Path("packaging_artifacts")
    tracker_app = packaging_root / "tracker_app"
    tracker_app.mkdir(parents=True, exist_ok=True)
    (tracker_app / "__init__.py").write_text("", encoding="utf-8")
    (tracker_app / "__main__.py").write_text(
        "def main() -> None:\n    print('tracker app launched')\n\nif __name__ == '__main__':\n    main()\n",
        encoding="utf-8",
    )
    (packaging_root / "requirements-dev.txt").write_text("pytest\n", encoding="utf-8")
    (packaging_root / "README.md").write_text("# Tracker App\n\n## Run\npython -m tracker_app\n", encoding="utf-8")


def main() -> None:
    service = TaskWorkflowService()
    service.add_unique_task("task-1", "Plan sprint")
    service.add_unique_task("task-2", "Write release report")
    _ = service.search("missing")
    write_artifacts(service)
    print('Hour 45 | test runner: pytest')
    print('Hour 45 | fixture name: sample_tasks')
    print('Hour 45 | passing unit tests: 3')
    print('Hour 45 | failed blank title case: caught')
    print('Hour 45 | testing rule: assert behavior, not implementation details')
    print('Hour 46 | coverage target: service layer')
    print('Hour 46 | edge case tested: duplicate id rejected')
    print('Hour 46 | integration test: json round trip')
    print('Hour 46 | regression guard: search empty result')
    print('Hour 46 | quality rule: cover the risky paths first')
    print('Hour 47 | requirements file: ready')
    print('Hour 47 | app entrypoint: python -m tracker_app')
    print('Hour 47 | readme run steps: documented')
    print('Hour 47 | optional executable: skipped')
    print('Hour 47 | packaging rule: ship the simplest runnable form first')
    print('Hour 48 | final demo domain: tracker')
    print('Hour 48 | final demo layers: ui, service, data')
    print('Hour 48 | review topics: oop, gui, db, api, analytics, testing')
    print('Hour 48 | retrospective note: keep separation of concerns')
    print('Hour 48 | final result: advanced capstone ready')


if __name__ == "__main__":
    main()
```
