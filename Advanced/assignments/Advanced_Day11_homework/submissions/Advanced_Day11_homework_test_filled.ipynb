## Part 1: Hour 41 - Pandas essentials: loading, cleaning, summarizing

**Goal:** Build a small cleaned DataFrame with labeled summary outputs.

**Best practice:** Normalize and clean the dataset before trusting summary numbers.

**Avoid this pitfall:** Summarizing dirty data hides missing or malformed values.


````python
from pathlib import Path
import pandas as pd

def get_assignment_root() -> Path:
    current = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
    return current.parent if current.name == "submissions" else current

assignment_root = get_assignment_root()
reports_dir = assignment_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

task_rows = [
    {"owner": "Ava", "priority": "low", "estimate_hours": 2.0, "actual_hours": 3.0},
    {"owner": "Ben", "priority": None, "estimate_hours": 3.0, "actual_hours": 3.8},
    {"owner": "Ava", "priority": "medium", "estimate_hours": 4.0, "actual_hours": 4.6},
    {"owner": "Cara", "priority": "high", "estimate_hours": 5.0, "actual_hours": 5.4},
    {"owner": "Ben", "priority": "medium", "estimate_hours": 4.0, "actual_hours": 4.6},
]
task_df = pd.DataFrame(task_rows)
task_df["priority"] = task_df["priority"].fillna("medium")
mean_estimate = task_df["estimate_hours"].mean()
owner_groups = task_df.groupby("owner").size()

print(f"Hour 41 | dataframe rows: {len(task_df)}")
print("Hour 41 | missing priority filled: yes")
print(f"Hour 41 | mean estimate hours: {mean_estimate:.1f}")
print(f"Hour 41 | grouped owners: {len(owner_groups)}")
print("Hour 41 | data rule: clean before you summarize")
````

## Part 2: Hour 42 - Visualization with matplotlib: charts you can ship

**Goal:** Build a saved matplotlib chart that supports the tracker report.

**Best practice:** Ship charts with clear labels, titles, and a saved artifact path.

**Avoid this pitfall:** A chart without labels looks finished but communicates very little.


````python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

priority_counts = task_df["priority"].value_counts().reindex(["low", "medium", "high"], fill_value=0)
chart_path = reports_dir / "priority_chart.png"
fig, ax = plt.subplots(figsize=(4, 3))
ax.bar(priority_counts.index.tolist(), priority_counts.tolist(), color=["#4c78a8", "#f58518", "#54a24b"])
ax.set_title("Task Count by Priority")
ax.set_xlabel("priority")
ax.set_ylabel("count")
fig.tight_layout()
fig.savefig(chart_path, format="png", dpi=120)
plt.close(fig)

print("Hour 42 | chart type: bar")
print("Hour 42 | x labels: low, medium, high")
print("Hour 42 | figure title: Task Count by Priority")
print("Hour 42 | saved file: reports/priority_chart.png")
print("Hour 42 | viz rule: label axes and title clearly")
````

## Part 3: Hour 43 - Regression demo (practical, limited scope)

**Goal:** Build a simple regression example with one labeled prediction.

**Best practice:** Keep the regression demo practical and explain it as a limited forecast.

**Avoid this pitfall:** Treating one quick regression as production-ready analytics overstates the result.


````python
from sklearn.linear_model import LinearRegression
import pandas as pd

regression_model = LinearRegression()
regression_model.fit(task_df[["estimate_hours"]], task_df["actual_hours"])
prediction = float(regression_model.predict(pd.DataFrame({"estimate_hours": [5.0]}))[0])

print("Hour 43 | model: linear regression")
print("Hour 43 | feature: estimate_hours")
print("Hour 43 | target: actual_hours")
print(f"Hour 43 | prediction for 5h estimate: {prediction:.1f}")
print("Hour 43 | scope rule: explain the demo, do not oversell it")
````

## Part 4: Hour 44 - Integrate analytics into capstone (report feature)

**Goal:** Build a report feature that connects data outputs to the tracker project.

**Best practice:** Fold summaries, charts, and predictions into one readable report feature.

**Avoid this pitfall:** Leaving analytics as disconnected notebook snippets weakens the capstone.


````python
summary_path = reports_dir / "weekly_summary.txt"
mean_actual = task_df["actual_hours"].mean()
summary_lines = [
    "Task Count by Priority",
    f"mean actual hours: {mean_actual:.1f}",
    f"next task hours: {prediction:.1f}",
    "included chart: priority_chart.png",
]
summary_path.write_text(chr(10).join(summary_lines), encoding="utf-8")

print("Hour 44 | report file: reports/weekly_summary.txt")
print("Hour 44 | included chart: priority_chart.png")
print("Hour 44 | included metric: mean actual hours")
print("Hour 44 | included prediction: next task hours")
print("Hour 44 | report rule: tie analytics back to capstone decisions")
````
