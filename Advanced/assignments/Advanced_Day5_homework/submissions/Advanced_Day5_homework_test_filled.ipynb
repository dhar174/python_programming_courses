## Part 1: Hour 17 - GUI fundamentals (Tkinter or ttk): event loop, widgets, callbacks

**Goal:** Build a simple Tkinter window with one working action.

**Best practice:** Create widgets clearly and wire callbacks to small handler functions.

**Avoid this pitfall:** Burying business rules directly in callbacks makes the GUI brittle.


```python
class FakeButton:
    def __init__(self, text: str, command) -> None:
        self.text = text
        self.command = command

class TrackerAppWindow:
    def __init__(self) -> None:
        self.title = "tracker app"
        self.widgets = ["label", "entry"]
        self.button = FakeButton("Add", self.add_task)
        self.widgets.append("button")
        self.last_action = ""

    def add_task(self) -> str:
        self.last_action = "add_task"
        return self.last_action

    def start(self) -> str:
        self.button.command()
        return "started"

tracker_window = TrackerAppWindow()
loop_state = tracker_window.start()

print(f"Hour 17 | root window: {tracker_window.title}")
print(f"Hour 17 | main widgets: {', '.join(tracker_window.widgets)}")
print(f"Hour 17 | event loop: {loop_state}")
print(f"Hour 17 | callback wired: {tracker_window.last_action}")
print("Hour 17 | gui rule: keep handlers thin")
```

## Part 2: Hour 18 - Layout and usability: grid, pack, spacing, resizing

**Goal:** Build a form layout that stays readable when the window changes size.

**Best practice:** Choose one layout strategy per container and test resizing early.

**Avoid this pitfall:** Mixing pack and grid in the same container causes layout bugs.


```python
layout_config = {
    "manager": "grid",
    "padding": 8,
    "column_stretch": True,
    "resize_behavior": "stable",
}
assert layout_config["padding"] == 8

print(f"Hour 18 | layout manager: {layout_config['manager']}")
print(f"Hour 18 | padding applied: {layout_config['padding']}")
print(f"Hour 18 | column stretch: {'enabled' if layout_config['column_stretch'] else 'disabled'}")
print(f"Hour 18 | resize behavior: {layout_config['resize_behavior']}")
print("Hour 18 | usability rule: align labels and inputs consistently")
```

## Part 3: Hour 19 - Forms, validation, and user feedback patterns

**Goal:** Build a validated form that reports success and error states clearly.

**Best practice:** Validate user input near the form boundary and show friendly feedback.

**Avoid this pitfall:** Failing silently after bad input leaves users stuck.


```python
class TaskForm:
    def __init__(self) -> None:
        self.status_label = ""

    def submit(self, title: str, priority: str) -> tuple[str, str]:
        if not title.strip():
            self.status_label = "blank title rejected"
            return self.status_label, self.status_label
        if priority not in {"low", "medium", "high"}:
            self.status_label = "invalid priority: warned"
            return title, self.status_label
        self.status_label = "task added"
        return title, self.status_label

task_form = TaskForm()
task_form.submit("   ", "medium")
valid_field, valid_status = task_form.submit("title entry", "medium")
_, invalid_priority_status = task_form.submit("title entry", "urgent")

print(f"Hour 19 | field read: {valid_field}")
print("Hour 19 | blank title: rejected")
print(f"Hour 19 | status label: {valid_status}")
print(f"Hour 19 | invalid priority: {invalid_priority_status.split(': ', 1)[-1]}")
print("Hour 19 | feedback rule: tell users what to fix")
```

## Part 4: Hour 20 - Record list UI with Listbox or Treeview and selection events

**Goal:** Build a list-based record browser with reliable selection handling.

**Best practice:** Bind selection events to stable record ids rather than row positions.

**Avoid this pitfall:** Using display order as identity makes updates and deletes unsafe.


```python
class RecordBrowser:
    def __init__(self, rows: list[dict[str, str]]) -> None:
        self.rows = rows
        self.selected_id = ""
        self.detail_panel = ""
        self.widget_name = "treeview"

    def select(self, task_id: str) -> None:
        self.selected_id = task_id
        self.detail_panel = next(row["title"] for row in self.rows if row["task_id"] == task_id)

record_browser = RecordBrowser(
    [
        {"task_id": "task-001", "title": "Plan sprint"},
        {"task_id": "task-002", "title": "Ship report"},
        {"task_id": "task-003", "title": "Review notes"},
    ]
)
record_browser.select("task-002")

print(f"Hour 20 | list widget: {record_browser.widget_name}")
print(f"Hour 20 | rows loaded: {len(record_browser.rows)}")
print(f"Hour 20 | selected id: {record_browser.selected_id}")
print("Hour 20 | detail panel: synced")
print("Hour 20 | selection rule: use stable record ids")
```
