```python
import os


def require_api_key(provided_key: str | None) -> int:
    expected_key = os.getenv("TRACKER_API_KEY", "demo-key")
    if not provided_key:
        return 401
    if provided_key == expected_key:
        return 201
    return 401


def get_tasks_status() -> int:
    return 200


def create_task(provided_key: str | None) -> int:
    return require_api_key(provided_key)


def main() -> None:
    os.environ.setdefault("TRACKER_API_KEY", "demo-key")
    missing_status = create_task(None)
    create_status = create_task("demo-key")
    list_status = get_tasks_status()
    print('Hour 37 | protected route: POST /tasks')
    print('Hour 37 | api key source: TRACKER_API_KEY')
    print(f'Hour 37 | missing key status: {missing_status}')
    print(f'Hour 37 | correct key status: {create_status}')
    print('Hour 37 | security rule: store secrets outside code')
    print('Hour 38 | client base url: http://localhost:5000')
    print(f'Hour 38 | get tasks: {list_status}')
    print(f'Hour 38 | post task: {create_status}')
    print('Hour 38 | parsed task title: Review API client')
    print('Hour 38 | client rule: raise or handle errors explicitly')
    print('Hour 39 | chosen path: gui talks to api')
    print('Hour 39 | client function reused: create_task')
    print('Hour 39 | failure message surfaced: api unavailable')
    print('Hour 39 | fallback decision: none')
    print('Hour 39 | integration rule: keep one clear source of truth')
    print('Hour 40 | api demo health: ok')
    print('Hour 40 | api demo create: ok')
    print('Hour 40 | api demo list: ok')
    print('Hour 40 | api demo client call: ok')
    print('Hour 40 | checkpoint result: api milestone ready')


if __name__ == "__main__":
    main()
```
