## Part 1: Hour 37 - API security basics: API key gate + safe storage

**Goal:** Build a simple API key check around write operations.

**Best practice:** Gate write operations with an environment-based API key in this course scope.

**Avoid this pitfall:** Hardcoding credentials in source or logs undermines the whole exercise.


```python
# Filled fixture answer for Part 1: Hour 37 - API security basics: API key gate + safe storage
import os

TRACKER_TASKS = [{"id": "task-1", "title": "Review API client"}]

def require_api_key(provided_key: str | None) -> int:
    expected_key = os.getenv("TRACKER_API_KEY", "demo-key")
    if not provided_key:
        return 401
    if provided_key == expected_key:
        return 201
    return 401

def get_tasks_status() -> int:
    return 200

def create_task(provided_key: str | None) -> int:
    return require_api_key(provided_key)

os.environ.setdefault("TRACKER_API_KEY", "demo-key")
missing_status = create_task(None)
create_status = create_task("demo-key")

print("Hour 37 | protected route: POST /tasks")
print("Hour 37 | api key source: TRACKER_API_KEY")
print(f"Hour 37 | missing key status: {missing_status}")
print(f"Hour 37 | correct key status: {create_status}")
print("Hour 37 | security rule: store secrets outside code")
```

## Part 3: Hour 39 - Integration choice: GUI talks to API OR parallel UI/API

**Goal:** Build executable integration behavior that keeps the API as the source of truth.

**Definition of Done (used by grading):**
1. Implement `submit_task_via_api(api_client, title)`.
2. Success path returns `status=201` and `source='api'`.
3. Failure path returns `status=503` with an `api unavailable` message.
4. Write `integration_artifacts/hour39_integration_trace.json` for success + failure attempts.
            


```python
import json
from pathlib import Path


def submit_task_via_api(api_client, title: str) -> dict:
    try:
        response = api_client.create_task(title)
        status = int(response.get("status", 201)) if isinstance(response, dict) else 201
        return {"status": status, "source": "api", "message": "created", "fallback": "none"}
    except ConnectionError as exc:
        return {"status": 503, "source": "api", "message": f"api unavailable: {exc}", "fallback": "none"}


class DemoSuccessClient:
    def create_task(self, title: str) -> dict:
        return {"status": 201, "task": {"title": title}}


class DemoFailingClient:
    def create_task(self, title: str) -> dict:
        raise ConnectionError("api unavailable")


integration_trace = [
    {"step": "success_create", **submit_task_via_api(DemoSuccessClient(), "Review API client")},
    {"step": "failure_create", **submit_task_via_api(DemoFailingClient(), "Review API client")},
]

artifact_dir = Path("integration_artifacts")
artifact_dir.mkdir(parents=True, exist_ok=True)
(artifact_dir / "hour39_integration_trace.json").write_text(json.dumps(integration_trace, indent=2), encoding="utf-8")

print(f"Hour 39 | success status: {integration_trace[0]['status']}")
print(f"Hour 39 | failure status: {integration_trace[1]['status']}")
            
```
