## Part 1: Hour 21 - CRUD wiring in GUI: update and delete workflows

**Goal:** Build GUI update and delete flows that round-trip through application state.

**Best practice:** Refresh the GUI from the current source of truth after each update or delete.

**Avoid this pitfall:** Mutating stale widget state without reloading causes mismatched displays.


```python
task_rows = [
    {"task_id": "task-001", "priority": "low"},
    {"task_id": "task-002", "priority": "medium"},
    {"task_id": "task-003", "priority": "high"},
]
selected_id = "task-002"

for row in task_rows:
    if row["task_id"] == selected_id:
        row["priority"] = "high"
task_rows = [row for row in task_rows if row["task_id"] != "task-003"]
list_refreshed = True

print(f"Hour 21 | selected id: {selected_id}")
print("Hour 21 | update action: priority set to high")
print("Hour 21 | delete action: task-003 removed")
print(f"Hour 21 | list refresh: {'complete' if list_refreshed else 'pending'}")
print("Hour 21 | crud rule: reload the view after state changes")
```

## Part 2: Hour 22 - GUI architecture: controller separation + avoiding callback spaghetti

**Goal:** Build a controller-style GUI structure with clean responsibilities.

**Best practice:** Keep controller methods thin and delegate business logic to services.

**Avoid this pitfall:** Callback spaghetti grows when widget code and domain rules live in the same function.


```python
class TaskService:
    def save(self, title: str) -> str:
        return f"saved {title}"

class TrackerApp:
    def __init__(self, service: TaskService) -> None:
        self.service = service
        self.current_selection_id = "task-002"
        self.widgets = {"save_button": "Save", "title_entry": "Entry"}

    def handle_save(self, title: str) -> str:
        return self.service.save(title)

tracker_app = TrackerApp(TaskService())
delegated_result = tracker_app.handle_save("Ship report")
assert delegated_result == "saved Ship report"

print(f"Hour 22 | ui class: {tracker_app.__class__.__name__}")
print("Hour 22 | service call: delegated")
print("Hour 22 | widget lookup: local to ui")
print(f"Hour 22 | state field: current_selection_id")
print("Hour 22 | architecture rule: callbacks coordinate, services decide")
```
